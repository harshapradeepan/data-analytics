```python
df['Age_mean_imputed'] = df['Age'].fillna(df['Age'].mean())
df['Age_median_imputed'] = df['Age'].fillna(df['Age'].median())
df
```

```python
 import matplotlib.pyplot as plt
 import seaborn as sns

 cols = ['Age', 'SibSp' , 'Parch' , 'Fare']

plt.figure(figsize=(12,8))

for i,col in enumerate(cols,1):
  plt.subplot(2,2,i)
  sns.boxplot(y=df[col])
  plt.title(f'Boxplot of {col}')
plt.tight_layout()
plt.show()
```

```python
df_clean = df.copy()
for coll in cols:

  Q1=df_clean[col].quantile(0.25)
  Q3=df_clean[col].quantile(0.75)

  IQR=Q3-Q1
  lower_limit=Q1-1.5*IQR
  upper_limit=Q3+1.5*IQR

  df_clean=df_clean[
      (df_clean[col] >= lower_limit) & (df_clean[col] <= upper_limit)
      ]
  print("Original shape:",df.shape)
  print("After outlier removal",df_clean.shape)
```

```python
plt.figure(figsize = (12,8))

for i , col in enumerate(cols,1):
    plt.subplot(2,2,i)
    sns.boxplot(y = df_clean[col])
    plt.title(f'{col} - After outlier removal')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

file_path = '/content/titanic uncleaned.csv'
titanic_data = pd.read_csv(file_path)

encoded_data = titanic_data.copy()

label_encoder = LabelEncoder()
encoded_data["Sex_Label"] = label_encoder.fit_transform(encoded_data['Sex'])

one_hot_encode = pd.get_dummies(encoded_data['Embarked'], prefix = 'Embarked')
encoded_data = pd.concat([encoded_data, one_hot_encode], axis=1)

pclass_mapping = {1: 'First', 2:'Second', 3:'Third'}

encoded_data['Pclass_Category'] = encoded_data['Pclass'].map(pclass_mapping)

print(encoded_data[['Sex','Sex_Label', 'Embarked','Embarked_Q','Embarked_S','Embarked_C','Pclass','Pclass_Category',]])
```
